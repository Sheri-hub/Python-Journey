```python
from Bio.Seq import Seq
```

```python
dna_seq = Seq("ATGCGATCGTACGATCGTAGCTAGCTAGCTGACT")
```

```python
def gc_content(seq):
    gc_count = seq.count("G") + seq.count("C")
    return (gc_count / len(seq)) * 100
```

```python
print(f"GC content: {gc_content(dna_seq):.2f}%")
```

```python
motif = "CGT"
positions = [i for i in range(len(dna_seq)) if dna_seq.startswith(motif, i)]
```

```python
print(f"Motif '{motif}' found at positions: {positions}")
```
